```python
# Create a dictionary of column names and their duplicate counts
dup_summary = {col: nmd.duplicated(subset=[col]).sum() for col in nmd.columns}

# Filter out columns that have 0 duplicates for a cleaner list
only_dupes = {k: v for k, v in dup_summary.items() if v > 0}

print("Columns with duplicates and their counts:")
print(only_dupes)
```
